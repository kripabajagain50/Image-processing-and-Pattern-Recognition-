```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab.patches import cv2_imshow

%matplotlib inline

# 1. Load and prepare image
img = cv2.imread('Lenna.jpg', 0) / 255.0  # Read as grayscale and normalize

# 2. Display original image
print("=== Original Image ===")
cv2_imshow(img * 255)  # Convert back to 0-255 range for display

# 3. Generate Gaussian noise
noise_intensity = 0.1  # Standard deviation of the noise
noise = np.random.normal(loc=0, scale=noise_intensity, size=img.shape)

# 4. Display the noise
print("\n=== Gaussian Noise ===")
cv2_imshow((noise * 255).clip(0, 255))

# 5. Add noise to image and clip values
noisy_img = np.clip(img + noise, 0, 1)  # Ensure values stay between 0-1

# 6. Display noisy image
print("\n=== Noisy Image ===")
cv2_imshow(noisy_img * 255)

# 7. Denoising Methods
# --- Gaussian Blur ---
gaussian_denoised = cv2.GaussianBlur((noisy_img * 255).astype(np.uint8), (5, 5), 1.5) / 255.0

# --- Median Blur ---
median_denoised = cv2.medianBlur((noisy_img * 255).astype(np.uint8), 5) / 255.0

# --- Non-Local Means Denoising (Best for Gaussian Noise) ---
nlm_denoised = cv2.fastNlMeansDenoising((noisy_img * 255).astype(np.uint8), None, h=10, templateWindowSize=7, searchWindowSize=21) / 255.0

# 8. Display denoised images
print("\n=== Gaussian Blur Denoised ===")
cv2_imshow(gaussian_denoised * 255)

print("\n=== Median Blur Denoised ===")
cv2_imshow(median_denoised * 255)

print("\n=== Non-Local Means Denoised ===")
cv2_imshow(nlm_denoised * 255)

# 9. Show all images together for comparison
plt.figure(figsize=(15, 10))

plt.subplot(2, 3, 1)
plt.imshow(img, cmap='gray', vmin=0, vmax=1)
plt.title('Original Image')
plt.axis('off')

plt.subplot(2, 3, 2)
plt.imshow(noise, cmap='gray')
plt.title(f'Gaussian Noise (σ={noise_intensity})')
plt.axis('off')

plt.subplot(2, 3, 3)
plt.imshow(noisy_img, cmap='gray', vmin=0, vmax=1)
plt.title('Noisy Image')
plt.axis('off')

plt.subplot(2, 3, 4)
plt.imshow(gaussian_denoised, cmap='gray', vmin=0, vmax=1)
plt.title('Gaussian Blur Denoised')
plt.axis('off')

plt.subplot(2, 3, 5)
plt.imshow(median_denoised, cmap='gray', vmin=0, vmax=1)
plt.title('Median Blur Denoised')
plt.axis('off')

plt.subplot(2, 3, 6)
plt.imshow(nlm_denoised, cmap='gray', vmin=0, vmax=1)
plt.title('Non-Local Means Denoised')
plt.axis('off')

plt.tight_layout()
```
